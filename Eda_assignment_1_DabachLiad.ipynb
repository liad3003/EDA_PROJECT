```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Set styling for professional-looking charts
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# ==============================================================================
# 1. LOAD DATASET & META-ANALYSIS (Sections 2 & 3)
# ==============================================================================
print("==================================================")
print("=== Section 3: Meta-Analysis & Data Structure  ===")
print("==================================================")

file_path = 'results.csv'

# Load dataset with fallback encodings to handle special characters
try:
    df = pd.read_csv(file_path, encoding='latin1')
    print("Dataset successfully loaded with 'latin1' encoding!\n")
except Exception:
    df = pd.read_csv(file_path, encoding='iso-8859-1')
    print("Dataset successfully loaded with fallback 'iso-8859-1' encoding!\n")

# Clean column names (strip unexpected spaces)
df.columns = df.columns.str.strip()

# Automatically detect date column variations and rename to 'Date'
date_cols = [c for c in df.columns if 'date' in c.lower()]
if date_cols:
    df.rename(columns={date_cols[0]: 'Date'}, inplace=True)

# Dataset dimensions and info
print(f"Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns\n")

print("--- Data Types & General Info ---")
df.info()

print("\n--- First 5 Rows ---")
display(df.head())

# Convert Date column to DatetimeIndex format if present
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
    df = df.sort_values('Date').reset_index(drop=True)
    print("\nDate column successfully converted to datetime format and sorted chronologically.")


# ==============================================================================
# 2. DATA QUALITY & COMPLETENESS (Section 4)
# ==============================================================================
print("\n==================================================")
print("=== Section 4: Data Quality & Completeness     ===")
print("==================================================")

# Missing Values Analysis
missing_counts = df.isnull().sum()
missing_percent = (missing_counts / len(df)) * 100
missing_df = pd.DataFrame({'Missing_Count': missing_counts, 'Missing_Percentage (%)': missing_percent})

print("--- Missing Values Summary ---")
print(missing_df[missing_df['Missing_Count'] > 0])

# Duplicates Check
full_duplicates = df.duplicated().sum()
print(f"\nFull Duplicate Rows Count: {full_duplicates}")

# Cardinality Analysis
print("\n--- Column Cardinality (Unique Values Count) ---")
cardinality = df.nunique()
print(cardinality)

# Zero-Variance Column Check
zero_var_cols = cardinality[cardinality <= 1].index.tolist()
if zero_var_cols:
    print(f"\nWarning - Columns with zero variance (single unique value): {zero_var_cols}")


# ==============================================================================
# 3. UNIVARIATE ANALYSIS & OUTLIERS (Section 5)
# ==============================================================================
print("\n==================================================")
print("=== Section 5: Univariate Analysis & Outliers   ===")
print("==================================================")

# 5.1 Numerical Variables Statistics & Outliers
numerical_cols = ['FTHG', 'FTAG', 'HS', 'AS']
available_num_cols = [col for col in numerical_cols if col in df.columns]

for col in available_num_cols:
    print(f"\n--- Detailed Statistics for: {col} ---")
    series = df[col].dropna()
    desc = series.describe()

    iqr = desc['75%'] - desc['25%']
    skew = series.skew()
    mad = (series - desc['50%']).abs().median()  # Median Absolute Deviation

    print(f"Mean: {desc['mean']:.2f} | Median: {desc['50%']:.2f} | Std: {desc['std']:.2f}")
    print(f"Min: {desc['min']:.0f} | Max: {desc['max']:.0f} | IQR: {iqr:.2f} | MAD: {mad:.2f}")
    print(f"Skewness: {skew:.2f}")

    # Outlier Method 1: Z-Score (> 3 Std)
    z_scores = (series - desc['mean']) / desc['std']
    outliers_z = series[np.abs(z_scores) > 3]

    # Outlier Method 2: IQR Method (1.5 * IQR)
    lower_bound = desc['25%'] - 1.5 * iqr
    upper_bound = desc['75%'] + 1.5 * iqr
    outliers_iqr = series[(series < lower_bound) | (series > upper_bound)]

    # Outlier Method 3: Quantiles Method (1st & 99th Percentile)
    q01 = series.quantile(0.01)
    q99 = series.quantile(0.99)
    outliers_q = series[(series < q01) | (series > q99)]

    print(f"Outliers detected -> Z-Score: {len(outliers_z)} | IQR: {len(outliers_iqr)} | Quantiles: {len(outliers_q)}")

# 5.2 Categorical Variables Analysis
categorical_cols = ['FTR', 'HomeTeam', 'Referee']
available_cat_cols = [col for col in categorical_cols if col in df.columns]

for col in available_cat_cols:
    print(f"\n--- Top Categories for '{col}' ---")
    counts = df[col].value_counts()
    percentages = df[col].value_counts(normalize=True) * 100
    display(pd.DataFrame({'Count': counts.head(5), 'Percentage (%)': percentages.head(5).round(2)}))


# ==============================================================================
# 4. CORRELATIONS & CATEGORICAL RELATIONSHIPS (Section 6.1 & 6.2)
# ==============================================================================
print("\n==================================================")
print("=== Section 6: Correlations & Categorical Binning ===")
print("==================================================")

num_metrics = ['FTHG', 'FTAG', 'HS', 'AS', 'HST', 'AST']
available_metrics = [col for col in num_metrics if col in df.columns]

if available_metrics:
    num_df = df[available_metrics].dropna()

    print("\n--- Pearson Correlation Matrix ---")
    display(num_df.corr(method='pearson').round(3))

    print("\n--- Spearman Correlation Matrix ---")
    display(num_df.corr(method='spearman').round(3))

    print("\n--- Kendall Correlation Matrix ---")
    display(num_df.corr(method='kendall').round(3))

# Categorical Crosstab & Binning
if 'HS' in df.columns and 'FTR' in df.columns:
    print("\n--- Categorical Relationships & Binning ---")
    shot_bins = [-1, 5, 10, 15, 100]
    shot_labels = ['Low (0-5)', 'Medium (6-10)', 'High (11-15)', 'Very High (16+)']
    df['home_shots_group'] = pd.cut(df['HS'], bins=shot_bins, labels=shot_labels)

    freq_table = pd.crosstab(df['home_shots_group'], df['FTR'], margins=True, margins_name='Total')
    print("Frequency Table: Home Shots Volume Group vs Match Outcome (FTR):")
    display(freq_table)


# ==============================================================================
# 5. VISUALIZATIONS (Section 6.3 - 8 Required Plots)
# ==============================================================================
print("\n==================================================")
print("=== Section 6.3: Visualizations & Data Story   ===")
print("==================================================")

# 1. Scatterplot: Home Shots vs Home Goals
if 'HS' in df.columns and 'FTHG' in df.columns:
    plt.figure(figsize=(10, 5))
    sns.scatterplot(data=df, x='HS', y='FTHG', alpha=0.4, color='darkblue')
    plt.title('Scatterplot: Home Shots vs. Home Goals', fontsize=14)
    plt.xlabel('Home Shots (HS)', fontsize=12)
    plt.ylabel('Home Goals (FTHG)', fontsize=12)
    plt.show()

# 2. Histogram: Total Match Goals Distribution
if 'FTHG' in df.columns and 'FTAG' in df.columns:
    df['total_goals'] = df['FTHG'] + df['FTAG']
    plt.figure(figsize=(10, 5))
    sns.histplot(df['total_goals'], bins=10, kde=True, color='skyblue', discrete=True)
    plt.title('Histogram: Distribution of Total Match Goals', fontsize=14)
    plt.xlabel('Total Goals in Match', fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    plt.show()

# 3. Barchart: Match Outcome Distribution (FTR)
if 'FTR' in df.columns:
    plt.figure(figsize=(10, 5))
    ftr_counts = df['FTR'].value_counts()
    sns.barplot(x=ftr_counts.index, y=ftr_counts.values, hue=ftr_counts.index, palette='viridis', legend=False)
    plt.title('Barchart: Full-Time Match Outcomes (H = Home Win, D = Draw, A = Away Win)', fontsize=14)
    plt.xlabel('Match Outcome (FTR)', fontsize=12)
    plt.ylabel('Match Count', fontsize=12)
    plt.show()

# 4. Boxplot: Home Shots on Target by Match Outcome
if 'HST' in df.columns and 'FTR' in df.columns:
    plt.figure(figsize=(10, 5))
    sns.boxplot(data=df, x='FTR', y='HST', hue='FTR', palette='Set2', legend=False)
    plt.title('Boxplot: Home Shots on Target (HST) by Match Outcome (FTR)', fontsize=14)
    plt.xlabel('Match Outcome (FTR)', fontsize=12)
    plt.ylabel('Home Shots on Target (HST)', fontsize=12)
    plt.show()

# 5. Violin Plot: Total Yellow Cards Distribution
if 'HY' in df.columns and 'AY' in df.columns and 'FTR' in df.columns:
    df['total_yellow_cards'] = df['HY'] + df['AY']
    plt.figure(figsize=(10, 5))
    sns.violinplot(data=df, x='FTR', y='total_yellow_cards', hue='FTR', palette='muted', legend=False)
    plt.title('Violin Plot: Total Yellow Cards per Match by Outcome', fontsize=14)
    plt.xlabel('Match Outcome (FTR)', fontsize=12)
    plt.ylabel('Total Yellow Cards', fontsize=12)
    plt.show()

# 6. Piechart: Proportion of Full-Time Match Outcomes
if 'FTR' in df.columns:
    plt.figure(figsize=(7, 7))
    ftr_counts = df['FTR'].value_counts()
    plt.pie(ftr_counts.values, labels=ftr_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
    plt.title('Piechart: Proportion of Match Outcomes', fontsize=14)
    plt.show()

# 7. Pairplot: Relationships between Key Metrics
pairplot_cols = ['FTHG', 'FTAG', 'HS', 'AS']
available_pairplot = [col for col in pairplot_cols if col in df.columns]

if len(available_pairplot) >= 3:
    sns.pairplot(df[available_pairplot].dropna(), corner=True)
    plt.suptitle('Pairplot: Key Match Metrics Relationships', y=1.02, fontsize=14)
    plt.show()

# 8. Heatmap: Pearson Correlation Matrix
if available_metrics:
    plt.figure(figsize=(8, 6))
    corr_matrix = df[available_metrics].dropna().corr(method='pearson')
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f", linewidths=0.5)
    plt.title('Heatmap: Pearson Correlation Matrix', fontsize=14)
    plt.show()


# ==============================================================================
# 6. INDEX ANALYSIS (Section 7)
# ==============================================================================
print("\n==================================================")
print("=== Section 7: Index Analysis                   ===")
print("==================================================")

print(f"Is Dataset Index Unique? {df.index.is_unique}")
print(f"Index Data Type: {type(df.index)}")

if 'Date' in df.columns and pd.api.types.is_datetime64_any_dtype(df['Date']):
    print(f"Dataset Date Range: From {df['Date'].min()} to {df['Date'].max()}")
    print(f"Is Sorted Chronologically? {df['Date'].is_monotonic_increasing}")


# ==============================================================================
# 7. BONUS: FEATURE ENGINEERING, HYPOTHESIS TESTING & TIME TRENDS (Section 9)
# ==============================================================================
print("\n=================================================================================")
print("=== Section 9: Bonus (Feature Engineering, Hypothesis Testing & Time Dependence) ===")
print("=================================================================================")

# Feature Engineering
if 'Date' in df.columns and pd.api.types.is_datetime64_any_dtype(df['Date']):
    df['Year'] = df['Date'].dt.year

if 'FTHG' in df.columns and 'FTAG' in df.columns:
    df['TotalGoals'] = df['FTHG'] + df['FTAG']
    df['GoalDiff'] = df['FTHG'] - df['FTAG']

if 'HS' in df.columns and 'HST' in df.columns:
    df['HomeShotAccuracy'] = np.where(df['HS'] > 0, df['HST'] / df['HS'], np.nan)

print("1. Feature Engineering Completed. First 5 rows of engineered features:")

# Display only available features safely to prevent KeyError
available_display_cols = [c for c in ['Date', 'HomeTeam', 'AwayTeam', 'TotalGoals', 'GoalDiff', 'HomeShotAccuracy'] if c in df.columns]
display(df[available_display_cols].head())

# Hypothesis Testing (Independent T-Test)
if 'HST' in df.columns and 'FTR' in df.columns:
    print("\n2. Hypothesis Testing (Independent T-Test):")
    home_win_shots = df[df['FTR'] == 'H']['HST'].dropna()
    non_home_win_shots = df[df['FTR'] != 'H']['HST'].dropna()

    t_stat, p_value = stats.ttest_ind(home_win_shots, non_home_win_shots)
    print(f"  -> Mean HST in Home Wins: {home_win_shots.mean():.2f}")
    print(f"  -> Mean HST in Non-Home Wins: {non_home_win_shots.mean():.2f}")
    print(f"  -> T-statistic: {t_stat:.4f}, P-value: {p_value:.4f}")
    if p_value < 0.05:
        print("  -> Conclusion: Statistically significant difference detected (p < 0.05).")

# Time Dependence Analysis
if 'Year' in df.columns and 'FTR' in df.columns:
    print("\n3. Generating Time Dependence Plot (Home Win % Trend)...")
    yearly_stats = df.groupby('Year').agg(
        total_matches=('FTR', 'count'),
        home_wins=('FTR', lambda x: (x == 'H').sum())
    ).reset_index()

    yearly_stats['home_win_pct'] = (yearly_stats['home_wins'] / yearly_stats['total_matches']) * 100
    yearly_stats = yearly_stats[yearly_stats['total_matches'] >= 50]

    plt.figure(figsize=(10, 5))
    sns.regplot(
        data=yearly_stats,
        x='Year',
        y='home_win_pct',
        scatter_kws={'alpha': 0.7, 'color': 'navy'},
        line_kws={'color': 'crimson', 'linewidth': 2}
    )
    plt.title("Time Dependence: Premier League Home Win Percentage Trend (1993–Present)", fontsize=13)
    plt.xlabel("Year", fontsize=11)
    plt.ylabel("Home Win Percentage (%)", fontsize=11)
    plt.show()
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("=================================================================================")
print("=== Section 9: Bonus (Feature Engineering, Hypothesis Testing & Time Dependence) ===")
print("=================================================================================")

# -------------------------------------------------------------------------
# 1. FEATURE ENGINEERING
# -------------------------------------------------------------------------
print("\n--- 1. Feature Engineering ---")

# Ensure Date is in datetime format and extract 'Year'
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
    df['Year'] = df['Date'].dt.year

# Create domain-specific features
if 'FTHG' in df.columns and 'FTAG' in df.columns:
    df['TotalGoals'] = df['FTHG'] + df['FTAG']
    df['GoalDiff'] = df['FTHG'] - df['FTAG']

if 'HS' in df.columns and 'HST' in df.columns:
    # Shot Accuracy Ratio (Avoid division by zero)
    df['HomeShotAccuracy'] = np.where(df['HS'] > 0, df['HST'] / df['HS'], np.nan)

print("Engineered Features Added: 'TotalGoals', 'GoalDiff', 'HomeShotAccuracy', 'Year'.")
print("First 5 rows of engineered metrics:")
display(df[['Date', 'HomeTeam', 'AwayTeam', 'FTHG', 'FTAG', 'TotalGoals', 'HomeShotAccuracy']].head())


# -------------------------------------------------------------------------
# 2. HYPOTHESIS TESTING
# -------------------------------------------------------------------------
print("\n--- 2. Hypothesis Testing ---")

# Hypothesis: Do Home Wins (FTR == 'H') have significantly higher Home Shots on Target (HST) than Non-Home Wins?
if 'HST' in df.columns and 'FTR' in df.columns:
    home_win_shots = df[df['FTR'] == 'H']['HST'].dropna()
    non_home_win_shots = df[df['FTR'] != 'H']['HST'].dropna()

    t_stat, p_value = stats.ttest_ind(home_win_shots, non_home_win_shots)

    print("Two-Sample T-Test: Comparing Home Shots on Target (HST) in Home Wins vs. Non-Home Wins")
    print(f"  -> Mean HST in Home Wins: {home_win_shots.mean():.2f}")
    print(f"  -> Mean HST in Non-Home Wins: {non_home_win_shots.mean():.2f}")
    print(f"  -> T-statistic: {t_stat:.4f}, P-value: {p_value:.4f}")

    if p_value < 0.05:
        print("  -> Conclusion: Statistically significant difference detected (p < 0.05). Home wins feature significantly higher shots on target.")
    else:
        print("  -> Conclusion: No statistically significant difference detected (p >= 0.05).")


# -------------------------------------------------------------------------
# 3. TIME DEPENDENCE & TREND ANALYSIS
# -------------------------------------------------------------------------
print("\n--- 3. Time Dependence Analysis ---")

if 'Year' in df.columns and 'FTR' in df.columns:
    # Group by year to calculate match volume and home win percentage
    yearly_stats = df.groupby('Year').agg(
        total_matches=('FTR', 'count'),
        home_wins=('FTR', lambda x: (x == 'H').sum()),
        avg_total_goals=('TotalGoals', 'mean')
    ).reset_index()

    yearly_stats['home_win_pct'] = (yearly_stats['home_wins'] / yearly_stats['total_matches']) * 100

    # Exclude years with incomplete match records (< 50 matches)
    yearly_stats = yearly_stats[yearly_stats['total_matches'] >= 50]

    # Plot 1: Home Win Percentage Trend over time
    plt.figure(figsize=(10, 5))
    sns.regplot(
        data=yearly_stats,
        x='Year',
        y='home_win_pct',
        scatter_kws={'alpha': 0.7, 'color': 'navy'},
        line_kws={'color': 'crimson', 'linewidth': 2}
    )
    plt.title("Time Dependence: Premier League Home Win Percentage Trend (1993–Present)", fontsize=13)
    plt.xlabel("Year", fontsize=11)
    plt.ylabel("Home Win Percentage (%)", fontsize=11)
    plt.tight_layout()
    plt.show()
```

* **Hypothesis Testing:** We performed an Independent Two-Sample T-Test to check whether matches resulting in a Home Win (`FTR == 'H'`) feature a significantly higher average number of Home Shots on Target (`HST`) compared to matches ending in a Draw or Away Win [1].
